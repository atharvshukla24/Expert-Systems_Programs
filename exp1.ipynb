8 puzzle problem using A* Algorithm

```python
class PuzzleState:
    def __init__(self, board, parent=None, move=None, depth=0, cost=0):
        self.board = board
        self.parent = parent
        self.move = move
        self.depth = depth
        self.cost = cost

    def __lt__(self, other):
        return self.cost < other.cost
```

```python
def printboard(board):
    for i in range(0, 9, 3):
        print(board[i:i+3])
```

```python
def move_tile(board, direction):
    board = board.copy()
    zero = board.index(0)
    row = zero // 3
    col = zero % 3

    if direction == "Up" and row > 0:
        new = zero - 3
    elif direction == "Down" and row < 2:
        new = zero + 3
    elif direction == "Left" and col > 0:
        new = zero - 1
    elif direction == "Right" and col < 2:
        new = zero + 1
    else:
        return None

    board[zero], board[new] = board[new], board[zero]

    return board
```

```python
def heuristic(board, goal):
    distance = 0

    for i in range(9):
        if board[i] != 0:
            goal_pos = goal.index(board[i])
            row1, col1 = divmod(i, 3)
            row2, col2 = divmod(goal_pos, 3)
            distance += abs(row1 - row2) + abs(col1 - col2)

    return distance
```

```python
def astar(start, goal):
    start_state = PuzzleState(
        start,
        cost=heuristic(start, goal)
    )

    queue = []
    heapq.heappush(queue, start_state)

    visited = set()

    while queue:
        current = heapq.heappop(queue)

        if tuple(current.board) in visited:
            continue

        visited.add(tuple(current.board))

        if current.board == goal:
            return current

        for direction in ["Up", "Down", "Left", "Right"]:
            new_board = move_tile(current.board, direction)

            if new_board is not None:
                if tuple(new_board) not in visited:
                    g = current.depth + 1
                    h = heuristic(new_board, goal)
                    f = g + h

                    new_state = PuzzleState(
                        new_board,
                        parent=current,
                        move=direction,
                        depth=g,
                        cost=f
                    )

                    heapq.heappush(queue, new_state)

    return None
```

```python
def print_solution(state):
    path = []

    while state:
        path.append(state)
        state = state.parent

    path.reverse()

    print("Solution Found")
    print("Number of moves:", len(path) - 1)

    for i, state in enumerate(path):
        print("\nStep", i)

        if state.move:
            print("Move:", state.move)

        printboard(state.board)
```

### Main Execution Block

```python
start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

print("Initial Board")
printboard(start)

print("\nGoal Board")
printboard(goal)

solution = astar(start, goal)

if solution:
    print_solution(solution)
else:
    print("No solution exists")
```
